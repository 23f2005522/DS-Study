## Types of Data Analysis
The three main types of data analysis are univariate, bivariate, and multivariate analysis

### Univariate Analysis
* **Variables:** Analyzes only a single variable at a time.
* **Objective:** Describe and summarize the data and find patterns within that single characteristic without considering causes or relationships.
* **Dependency:** No dependent variable.
* **Techniques & Measures:** Descriptive statistics including mean, median, mode, range, variance, standard deviation, minimum, maximum, and quartiles.
* **Visualization:** Histograms, bar charts, pie charts, box plots, and frequency distribution tables.
* **Example:** Studying the height or exam marks of students.

### Bivariate Analysis
* **Variables:** Involves two different variables to examine relationships.
* **Objective:** Explore the association, correlation, or causal relationship between two distinct quantities.
* **Dependency:** Can involve one dependent variable and one independent variable.
* **Techniques & Measures:** Correlation coefficients, covariance, cross-tabulation, and simple linear regression.
* **Visualization:** Scatter plots, regression lines, and grouped bar charts.
* **Example:** Comparing outdoor temperature against ice cream sales.

### Multivariate Analysis
* **Variables:** Analyzes three or more variables simultaneously.
* **Objective:** Understand complex relationships, interactions, and dependencies among multiple variables.
* **Dependency:** Can involve multiple dependent variables.
* **Techniques & Measures:** Multiple regression, Principal Component Analysis (PCA), factor analysis, cluster analysis, and MANOVA (Multivariate Analysis of Variance).
* **Visualization:** Heatmaps, pair plots, 3D scatter plots, and contour plots.
* **Example:** Predicting an employee's click rate or analyzing customer purchasing segments using ad type, gender, and income together.

### Connetion between types of analysis and plotting graphs
- *plotting graph for one column comes under univariate analysis*
- *plotting graph for two columns comes under bivariate analysis*
- *plotting graph for more than two columns comes under multivariate analysis*

```python
# from a pandas dataframe
batsman = pd.read_csv('../data/sharma-kohli.csv')
batsman

plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison in IPL')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=5) # solid, dashed, dashdot, dotted
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=1)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
## marker are the points (x_i,y_i) on the graph
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='o',markersize=10) # marker: o, s, D, ^, v, <, >, p, h
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='^')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend -> location 
# for labeling the lines in the graph lines  we can use label parameter in the plot function and then use plt.legend() to show the labels in the graph
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o',label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend() # loc = 'best', 'upper right', 'upper left', 'lower left', 'lower right', 'right', 'center left', 'center right', 'lower center', 'upper center', 'center'
```

```python
# limiting axes : from x_initial to x_final we can trim the graph using xlim and see the graph in that range only silimilarly for y-axis we can use ylim
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price , color='red' , marker="h" , linestyle="dashdot" )
plt.ylim(40000,60000) # y-axis limit
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid() # axis = 'x' or 'y' or 'both'
```

```python
# show :  is the final step to show the graph
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid()

plt.show()
```
