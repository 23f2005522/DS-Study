### Seaborn Roadmap

-   **figure-level** : it represents the entire figure and can contain multiple axes and figures can have multiple subplots
-   **axes-level** : it represents a single plot under the figure and can be used to create a single plot or multiple plots in a single figure

Types of Functions

- **Figure Level Functions** - They create a figure-level object and can contain multiple axes. Examples include `sns.relplot()`, `sns.catplot()`, and `sns.pairplot()`. They help in creating complex visualizations with multiple subplots and facets.
    - **Facet plot** : A facet plot (also called a "small-multiple" or trellis plot) splits a dataset into multiple subplots, where each panel displays a different subset of data based on one or more categorical variables.
- **Axis Level Functions** - They create an axes-level object and are used to create a single plot or multiple plots in a single figure. Examples include `sns.scatterplot()`, `sns.histplot()`, and `sns.countplot()`.

Main Classification

- Relational Plot

### 1. Relational Plot

- to see the statistical relation between 2 or more variables.
- `Bivariate Analysis` (means analyzing 2 variables at a time)

Plots under this section

- `scatterplot` : It is a type of plot that shows the relationship between two variables by displaying data points as dots on a two-dimensional graph. Each dot represents a single observation, and the position of the dot is determined by the values of the two variables being compared. Scatter plots are useful for identifying patterns, trends, and correlations between variables.
- `lineplot` : It is a type of plot that displays data points connected by straight lines. It is commonly used to visualize trends over time or to show the relationship between two continuous variables. Line plots are useful for identifying patterns, trends, and changes in data over a continuous range.

``````python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
sns.set_theme()
``````

``````python
tips = sns.load_dataset('tips')
tips
``````

``````python
sns.color_palette()
``````

``````python
# scatter plot -> axes level function
sns.scatterplot(data=tips, x='total_bill', y='tip',hue='sex',style='time',size='size' , palette='Set2') 

# `hue` is used to color the points based on the sex column
# `style` is used to change the marker style based on the time column
# `size` is used to change the size of the points based on the size column
``````

``````python
# relplot -> figure level -> square shape
## kind = "scatter" / "line"  

sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex',style='time',size='size')
``````

``````python
# line plot
gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'India']
temp_df
``````

``````python
# axes level function
sns.lineplot(data=temp_df, x='year', y='lifeExp' , marker='h' , color='#F54927' , label='Life Expectancy' , linestyle='--')
``````

``````python
# using relpplot
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line' , marker='o', color='#08B49D' , label='Life Expectancy' , linestyle='-')
``````

``````python
# hue -> style
temp_df = gap[gap['country'].isin(['India','Brazil','France'])]
temp_df
``````

``````python
sns.relplot(kind='line', data=temp_df, x='year', y='lifeExp', hue='country' , style="continent" , palette='Set1' , markers=True )

## notice a thing this is a figure level function thats whyb it is able to fit the ledgends outside the axes object area for reference see the below cell which is a axes level function and it is not able to fit the legends outside the axes object area and the ledgends are inside the axes object area
``````

``````python
sns.lineplot(data=temp_df, x='year', y='lifeExp', hue='country')
``````

``````python
## what is facet plot?
## facet plot is a way to visualize the relationship between only  ```two variables``` by creating multiple subplots based on the values of one or more categorical variables. Each subplot represents a subset of the data, allowing for easy comparison of patterns and trends across different groups. Facet plots are particularly useful for exploring interactions between variables and identifying differences in relationships across categories.

# facet plot -> `````works only with figure level function````` -> work with relplot which is a figure level function 
# it will not work with .scatterplot() and .lineplot()
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex', row='smoker')
``````

``````python
# col wrap
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=4)
``````

### 2. Distribution Plots 
-  These are the plots that are used to visualize the distribution of a single variable. They help in understanding the underlying distribution of the data, identifying patterns, and detecting outliers. Distribution plots can be used for both continuous and categorical variables.

    - used for univariate analysis
    - used to find out the distribution
    - Range of the observation
    - Central Tendency
    - is the data bimodal?
    - Are there outliers?

Plots under distribution plot

- **histplot** : `sns.histplot(df, x='x_column', hue='hue_column' , multiple='stack' , palette='palette_name')` or `sns.histplot(df, x='x_column', hue='hue_column' , multiple='dodge' , palette='palette_name')` or `sns.histplot(df, x='x_column', hue='hue_column' , multiple='fill' , palette='palette_name')`
- **kdeplot**: `sns.kdeplot(df, x='x_column', y='y_column' , hue='hue_column' , fill=True , palette='palette_name')` or `sns.kdeplot(df, x='x_column', y='y_column' , hue='hue_column' , fill=False , palette='palette_name')`
- **rugplot** : `sns.rugplot(df, x='x_column', hue='hue_column' , palette='palette_name')` or `sns.rugplot(df, x='x_column', hue='hue_column' , palette='palette_name')`

``````python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
``````

``````python
# plotting univariate histogram

sns.histplot(data=tips, x="total_bill" )
``````

``````python
sns.displot(data=tips, x='total_bill', kind='hist')
``````

``````python
# bins parameter
sns.displot(data=tips, x='total_bill', kind='hist',bins=2)
``````

``````python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables

# countplot
sns.displot(data=tips, x='day', kind='hist')
``````

``````python
# hue parameter
sns.displot(data=tips, x='tip', kind='hist',hue='sex')
``````

``````python
# element -> step
sns.displot(data=tips, x='tip', kind='hist',hue='sex',element='step')
``````

``````python
titanic = sns.load_dataset('titanic')
titanic
``````

``````python
sns.displot(data=titanic, x='age', kind='hist',element='step',col='sex' , hue="survived")
``````

``````python
# faceting using col and row -> not work on histplot function
sns.displot(data=tips, x='tip', kind='hist',col='sex',element='step')
``````

``````python
# kdeplot
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate
sns.kdeplot(data=tips,x='total_bill')
``````

``````python
sns.displot(data=tips,x='total_bill',kind='kde')
``````

``````python
# hue -> fill
sns.displot(
    data=tips, x="total_bill", kind="kde", hue="sex", fill=True, height=5, aspect=3
)
``````

``````python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
``````

``````python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips, x='total_bill', y='tip')
sns.displot(data=tips, x='total_bill', y='tip',kind='hist' )
``````

``````python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
sns.kdeplot(data=tips, x='total_bill', y='tip')
``````

### 2. Matrix Plot

- Heatmap
- Clustermap

``````python
# Heatmap

# Plot rectangular data as a color-encoded matrix
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')

# axes level function
plt.figure(figsize=(8,8))
sns.heatmap(temp_df)
``````

``````python
# annot
temp_df = gap[gap["continent"] == "Europe"].pivot(
    index="country", columns="year", values="lifeExp"
)

plt.figure(figsize=(10, 10))
sns.heatmap(temp_df, annot=True, linewidth=0.5, cmap="summer")
``````

``````python
# Clustermap

# Plot a matrix dataset as a hierarchically-clustered heatmap.

# This function requires scipy to be available.

iris = px.data.iris()
iris
``````

``````python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
``````
