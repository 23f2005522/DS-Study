# CONSIDERING THE WHOLE CONTAINER



```python
company=soup.find_all('div',class_='flex cursor-pointer px-5 pt-5 gap-2 max-md:px-3 max-md:pt-4')
```

```python
len(company) 
```

```python
name=[]
rating=[]
reviews=[]
domain=[]

for i in company:
  # print(i.find('h2').text.strip())
  # print(i.find('span',class_="text-primary-text font-pn-700 text-sm !text-[14px] !leading-[14px]").text.strip())
  # print(i.find('span' , class_='pl-1 text-[14px] font-pn-400 leading-5 text-[#4E597B] max-md:text-[12px]').text.strip("()"))
  # print(i.find_all('span',class_='overflow-hidden text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#394468] max-md:text-[12px] max-md:leading-4')[0].text.strip().split("|")[0])

  
  name.append(i.find('h2').text.strip())
  rating.append(float(i.find('span',class_="text-primary-text font-pn-700 text-sm !text-[14px] !leading-[14px]").text.strip()))
  reviews.append(i.find('span' , class_='pl-1 text-[14px] font-pn-400 leading-5 text-[#4E597B] max-md:text-[12px]').text.strip("()"))
  domain.append(i.find_all('span',class_='overflow-hidden text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#394468] max-md:text-[12px] max-md:leading-4')[0].text.strip().split("|")[0]) 
  

df=pd.DataFrame({'name':name,
   'rating':rating,
   'reviews':reviews,
    'domain':domain
   })

df

  
```
