# 9. Skiprows/nrows Parameter


- **skiprows** parameter is used to skip the rows from the top of the file. It can be an integer or a list of integers. If it is an integer, it will skip that many rows from the top. If it is a list of integers, it will skip those specific rows.
    - syntax: `skiprows = [0, 1, 2]` or `skiprows = 3`  means skip row number 0, 1, 2 or skip first 3 rows from the top of the file.
- **nrows** parameter is used to read only the specified number of rows from the file. It can be an integer. If it is an integer, it will read that many rows from the top.
    - syntax: `nrows = 5` means read only first 5 rows from the top of the file.

```python
pd.read_csv('./data/aug_train.csv',nrows=100)
```

# 11. Skip bad lines

```python
pd.read_csv('./data/BX-Books.csv', sep=';' ,on_bad_lines='skip')  ## arguments for on_bad_lines are 'error' , 'warn' , 'skip' . Default is 'error'. If we set it to 'warn' then it will show warning message for bad lines and if we set it to 'skip' then it will skip the bad lines without showing any warning message.
```

```python
def rename(name):
    
    ## using switch case
    short_name = {
        "Royal Challengers Bangalore": "RCB",
        "Sunrisers Hyderabad": "SRH",
        "Kings XI Punjab": "KXIP",
        "Delhi Daredevils": "DD",
        "Kolkata Knight Riders": "KKR",
        "Chennai Super Kings": "CSK",
        "Rajasthan Royals": "RR",
        "Mumbai Indians": "MI",
        "Punjab": "PB",
    }
    return short_name.get(name, name)
```

```python
rename("Royal Challengers Bangalore")
```

# 15. na_values parameter
- before even importing the csv file, we can specify the values which we want to treat as NaN values. For example, if we have a csv file with missing values represented as "xx", "x", or "-", we can use the `na_values` parameter to specify these values as NaN. 
- This will help in handling missing data effectively by using methods like `dropna()` or `fillna()` to clean or impute the missing values in the dataset.


```python
pd.read_csv('./data/nan.csv',na_values=[])
```

```python
pd.read_csv('./data/nan.csv',na_values=["xx","x","-"])
```
